```python
import torch

from matplotlib import pyplot as plt

class Neuron:

  def __init__(self, inputs, weights, bias, target):

       self.inputs = inputs.detach().clone()

       self.weights = weights.detach().clone()

       self.bias = float(bias)

       self.target = float(target)

       self.output = 0.0

  def activate(self):

      weighted_sum = torch.dot(self.inputs, self.weights)

      self.output = weighted_sum + self.bias

      return self.output

 # Calculating the error

  def loss(self, output):

      # Mean Squared Error

      return (self.target - output) ** 2

  def backpropagation(self, learning_rate=0.001):

        # Loss derivative

        loss_gradient = -2 * (self.target - self.output)

        # Updating weights

        gradient_weights = loss_gradient * self.inputs

        self.weights -= learning_rate * gradient_weights

        # Updating bias

        self.bias -= learning_rate * loss_gradient


# Neuron Creation

my_neuron = Neuron(
    inputs= torch.rand(3),
    weights= torch.rand(3),
    bias= torch.randint(0, 10, ()).item(),
    target=2.0,
)

no_of_epoches = int(input("How much epoches would you like?: "))

outputs = []

for epoch in range(no_of_epoches):

    activated = my_neuron.activate()

    current_loss = my_neuron.loss(activated)

    # Convert tensor to float for easier plotting
    outputs.append(float(my_neuron.output))

    my_neuron.backpropagation(learning_rate=0.01)

    print(f"Epoch {epoch} | Target {my_neuron.target} | Output {my_neuron.output:.4f} | Loss {current_loss:.4f}")

    # Check for convergence or the final epoch
    if epoch == no_of_epoches - 1 or abs(my_neuron.output - my_neuron.target) < 1e-6:

      plt.figure(figsize=(10, 5))
      plt.plot(outputs, label='Neuron Output')
      plt.axhline(y=my_neuron.target, color='r', linestyle='--', label='Target')
      plt.title("Neuron Training Progress")
      plt.xlabel("Epoch")
      plt.ylabel("Value")
      plt.legend()
      plt.show()

      break
```
