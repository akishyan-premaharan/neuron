```python
import torch

class Neuron:

  def __init__(self, inputs, weights, bias, target):

       self.inputs = torch.tensor(inputs)

       self.weights = torch.tensor(weights)

       self.bias = bias

       self.raw_target = target

       self.raw_output = 0.0

       # Sigmoid places the target on a range of 0-1.

       self.target = torch.sigmoid(torch.tensor(target))

       self.output = 0.0

  # Activating the neuron (Known as a forward pass)

  def activate(self):

      weighted_sum = sum(input * weight for input, weight in zip(self.inputs, self.weights))

      output = weighted_sum + self.bias

      self.output = torch.sigmoid(output)

      self.raw_output = output

      return self.output

 # Calculating the error

  def loss(self, output):

      # Mean absolute error

      return torch.abs((output - self.target)/2)

  def backpropagation(self, learning_rate=0.1):

        loss_derivative = 1.0 if self.output > self.target else -1.0

        sigmoid_derivative = self.output * (1 - self.output)

        # Chain rule (The derivative of multiple functions (Inputs, inputs x weights, adding bias, Mean Absolute Error))

        base_gradient = loss_derivative * sigmoid_derivative

        # Updating weights

        gradient_weights = base_gradient * self.inputs

        self.weights -= learning_rate * gradient_weights

        # You can perform arithmetic on PyTorch tensors ^

        # Updating bias

        self.bias -= learning_rate * base_gradient


# Neuron Creation

my_neuron = Neuron(
    inputs=[1.0, 2.0, 0.5],
    weights=[0.4, -0.5, 1.2],
    bias=0.8,
    target=2.0,
)

try:

  repeat = int(input("How many times do you want to repeat?: "))

except:

  repeat = 1000
  print("Invalid input, defaulting to 1000")


for epoch in range(repeat):

    result = my_neuron.activate()
    loss = my_neuron.loss(result)
    my_neuron.backpropagation()

    print(f"Epoch {epoch + 1} | Raw Output: {my_neuron.raw_output:.4f} | Target: {my_neuron.raw_target}")
```
